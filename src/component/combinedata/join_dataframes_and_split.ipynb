```python
import os
import pandas as pd
import geopandas as gpd
import random
```

```python
path = '/content/drive/MyDrive/ClipPolygon'

shp_files = [file for file in os.listdir(path) if file.endswith(".shp")]
random.shuffle(shp_files)
shp_train = [shp_files[i] for i in range(int(0.7*len(shp_files)))]
shp_valid = [shp_files[i] for i in range(int(0.7*len(shp_files)) , int(0.7*len(shp_files)) + round(0.15*len(shp_files)))]
shp_test = [shp_files[i] for i in range(int(0.7*len(shp_files)) + round(0.15*len(shp_files)) , int(0.7*len(shp_files)) + 2*round(0.15*len(shp_files)))]
print(shp_train)
print(shp_valid)
print(shp_test)
```

```python
df = gpd.read_file('/content/drive/MyDrive/lagos_data.geojson')
df = df.to_crs('epsg:4326')
```

```python
training_points = []
validation_points = []
testing_points = []
```

```python
def prepare_dataset(l, shp, df2):
     for file in shp:
          df1 = gpd.read_file(os.path.join(path, file))
          df1 = df1.to_crs('epsg:4326')
          merged_df = gpd.sjoin(df2, df1, how='inner', predicate='within')
          l.append(merged_df)
     return l
```

```python
training_set = prepare_dataset(training_points, shp_train, df)
validation_set = prepare_dataset(validation_points, shp_valid, df)
test_set = prepare_dataset(testing_points, shp_test, df)
print(training_set)
print(validation_set)
print(test_set)
```

```python
train_df = pd.concat(training_set, ignore_index=True)
validation_df = pd.concat(validation_set, ignore_index=True)
test_df = pd.concat(test_set, ignore_index=True)

train_df.to_csv(path + '/Train/train.csv', index=False)
validation_df.to_csv(path + '/Validation/validation.csv', index=False)
test_df.to_csv(path + '/Test/test.csv', index=False)
```
